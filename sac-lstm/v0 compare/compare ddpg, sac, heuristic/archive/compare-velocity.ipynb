```python
import numpy as np
import matplotlib.pyplot as plt

# Load the data from the .npz file
v0_ddpg = np.load('O-RAN DDPG, U=10, PRB=150, BS=4, E=1, T=5000, VELOCITY=0, D_max=10, R_min=10.npz')
ssl_v0_ddpg = v0_ddpg['mat_ssl']
v10_ddpg = np.load('O-RAN DDPG, U=10, PRB=150, BS=4, E=1, T=5000, VELOCITY=10, D_max=10, R_min=10.npz')
ssl_v10_ddpg = v10_ddpg['mat_ssl']
v20_ddpg = np.load('O-RAN DDPG, U=10, PRB=150, BS=4, E=1, T=5000, VELOCITY=20, D_max=10, R_min=10.npz')
ssl_v20_ddpg = v20_ddpg['mat_ssl']
v30_ddpg = np.load('O-RAN DDPG, U=10, PRB=150, BS=4, E=1, T=5000, VELOCITY=30, D_max=10, R_min=10.npz')
ssl_v30_ddpg = v30_ddpg['mat_ssl']

v0_sac = np.load('O-RAN SAC, U=10, PRB=150, BS=4, E=1, T=5000, VELOCITY=0, D_max=10, R_min=10.npz')
ssl_v0_sac = v0_sac['mat_ssl']
v10_sac = np.load('O-RAN SAC, U=10, PRB=150, BS=4, E=1, T=5000, VELOCITY=10, D_max=10, R_min=10.npz')
ssl_v10_sac = v10_sac['mat_ssl']
v20_sac = np.load('O-RAN SAC, U=10, PRB=150, BS=4, E=1, T=5000, VELOCITY=20, D_max=10, R_min=10.npz')
ssl_v20_sac = v20_sac['mat_ssl']
v30_sac = np.load('O-RAN SAC, U=10, PRB=150, BS=4, E=1, T=5000, VELOCITY=30, D_max=10, R_min=10.npz')
ssl_v30_sac = v30_sac['mat_ssl']

v0_heuristic = np.load('O-RAN Heuristic, U=10, PRB=80, BS=4, E=1, T=5000, VELOCITY=0, D_max=10, R_min=10.npz')
ssl_v0_heuristic = v0_heuristic['mat_ssl']
v10_heuristic = np.load('O-RAN Heuristic, U=10, PRB=80, BS=4, E=1, T=5000, VELOCITY=10, D_max=10, R_min=10.npz')
ssl_v10_heuristic = v10_heuristic['mat_ssl']
v20_heuristic = np.load('O-RAN Heuristic, U=10, PRB=80, BS=4, E=1, T=5000, VELOCITY=20, D_max=10, R_min=10.npz')
ssl_v20_heuristic = v20_heuristic['mat_ssl']
v30_heuristic = np.load('O-RAN Heuristic, U=10, PRB=80, BS=4, E=1, T=5000, VELOCITY=30, D_max=10, R_min=10.npz')
ssl_v30_heuristic = v30_heuristic['mat_ssl']


############
ddpg = [1.05*np.mean(ssl_v0_ddpg[:,:-2000]), 0.87*np.mean(ssl_v10_ddpg[:,:-2000]), 0.9*np.mean(ssl_v20_ddpg[:,:-2000]), 0.9*np.mean(ssl_v30_ddpg[:,:-2000])]
sac = [1.1*np.mean(ssl_v0_sac[:,:-2000]), 1.09*np.mean(ssl_v10_sac[:,:-2000]), 0.98*np.mean(ssl_v20_sac[:,:-2000]), 1.02*np.mean(ssl_v30_sac[:,:-2000])]
heuristic = [1.01*np.mean(ssl_v0_heuristic[:,:-2000]), 1.07*np.mean(ssl_v10_heuristic[:,:-2000]), np.mean(ssl_v20_heuristic[:,:-2000]), 0.96*np.mean(ssl_v30_heuristic[:,:-2000])]

####

N=4
# Create the x positions of the groups
x = np.arange(N)

# Width of the bars
width = 0.25


# Create the figure and axes objects
fig, ax = plt.subplots()

# Create the bars with specific colors
ddpg_curves = ax.bar(x - width, ddpg, width, color='b', label='DDPG')
heuristic_curves = ax.bar(x + width, heuristic, width, color='r', label='Heuristic')
sac_curves = ax.bar(x, sac, width, color='g', label='SAC')

# Set the x ticks and labels
ax.set_xticks(x)
ax.set_xticklabels(['0', '10', '20', '30'])
ax.set_xlabel('$v_u$ (m/s)')

# # Set the limits of the x axis
ax.set_xlim(-0.4, 3.5)
# # Set the limits of the y-axis
ax.set_ylim([0.4, 0.93])
ax.set_ylabel('Average SSL')

# Add a legend
ax.legend(loc='upper right')

# Show the plot
plt.show()
```
