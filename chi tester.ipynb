```python
import numpy as np

class RadioResourceAllocator:
    def __init__(self, USER_NO, BS_NO):
        self.USER_NO = USER_NO
        self.BS_NO = BS_NO
        self.H_b = np.zeros([BS_NO, USER_NO])  # Example channel gain matrix
        
    def update_channel_gains(self, H):
        for u in range(self.USER_NO):
            for b in range(self.BS_NO):
                self.H_b[b, u] = np.average(H[b, :, u])

    def user_association(self, action):
        # Assuming 'action' is now a flattened array of preferences for each BS per user
        preferences = np.reshape(action, (self.USER_NO, self.BS_NO))
        
        # Apply softmax to convert preferences to probabilities
        probabilities = np.exp(preferences) / np.sum(np.exp(preferences), axis=1, keepdims=True)
        
        # Multiply by channel gains to weigh by channel quality
        weighted_preferences = probabilities * self.H_b.T
        
        # Select the BS with the maximum weighted preference for each user
        self.chi_num = np.argmax(weighted_preferences, axis=1)
        
        # Create user-BS association matrix
        self.chi = np.zeros((self.USER_NO, self.BS_NO))
        self.chi[np.arange(self.USER_NO), self.chi_num] = 1
        
        return self.chi

# Example usage
allocator = RadioResourceAllocator(10, 5)  # 10 users, 5 base stations
allocator.update_channel_gains(H)  # Update channel gains somehow
allocator.user_association(action)  # Pass the action from your DRL model
```
