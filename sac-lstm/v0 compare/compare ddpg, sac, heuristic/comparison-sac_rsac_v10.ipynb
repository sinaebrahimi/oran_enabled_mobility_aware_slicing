```python
#SSL_u JFI
## Calculate proportional fairness score for individual bitrate for each Episode
E=1
T=4200
USER_NO=10
WINDOW_SIZE=300

# Function to calculate Jain's Fairness Index
def calculate_jfi(values):
    numerator = np.sum(values) ** 2
    denominator = len(values) * np.sum(values ** 2)
    if denominator == 0:
        return 0  # Avoid division by zero
    return numerator / denominator

# Calculate JFI for each episode and each time slot
jfi_values_sac = np.zeros((E, T))
jfi_values_ddpg = np.zeros((E, T))
jfi_values_heuristic = np.zeros((E, T))
for e in range(E):
    for t in range(T):
        rate_t_sac = rate_sac[e, :, t]
        jfi_values_sac[e, t] = calculate_jfi(rate_t_sac)

        rate_t_ddpg = rate_ddpg[e, :, t]
        jfi_values_ddpg[e, t] = calculate_jfi(rate_t_ddpg)

        rate_t_heuristic = rate_heuristic[e, :, t]
        jfi_values_heuristic[e, t] = calculate_jfi(rate_t_heuristic)

# Average the JFI values over all episodes for each time slot
average_jfi_values_sac = np.mean(jfi_values_sac, axis=0)
average_jfi_values_ddpg = np.mean(jfi_values_ddpg, axis=0)
average_jfi_values_heuristic = np.mean(jfi_values_heuristic, axis=0)

data = [moving_average(average_jfi_values_ddpg, WINDOW_SIZE), moving_average(average_jfi_values_sac, WINDOW_SIZE), moving_average(average_jfi_values_heuristic, WINDOW_SIZE)]
names = ['DDPG', 'SAC','Heuristic']
averages = [np.average(average_jfi_values_ddpg),np.average(average_jfi_values_sac), np.average(average_jfi_values_heuristic)]
labels = ['{}: {:.2f} (on average)'.format(n, a) for n, a in zip(names, averages)]

colors = ['b', 'g', 'r']  # choose colors for each curve
linestyles = ['-', '-', '-']  # choose line styles for each curve

plot_graph('fairness-ssl-u', data, labels, colors, linestyles, "Episode", "Fairness score") 
#Normalized Jain\'s fairness score
```

```python
############
## Calculate PRB utilization for each user at each Episode and for each MC run
prb_utilization_sac = np.sum(mat_rho_sac, axis=(0, 1))  # Sum over PRBs and BSs
prb_utilization_heuristic = np.sum(mat_rho_heuristic, axis=(0, 1))  # Sum over PRBs and BSs
prb_utilization_ddpg = np.sum(mat_rho_ddpg, axis=(0, 1))  # Sum over PRBs and BSs

## Calculate proportional fairness score for each Episode
E=1
T=5000
BS_NO=4
fairness_scores_sac = np.zeros((E, T))
fairness_scores_heuristic = np.zeros((E, T))
fairness_scores_ddpg = np.zeros((E, T))

for e in range(E):
    for t in range(T-1):
        # Calculate Jain's fairness index for Episode t in MC run mc
        sum_of_prbs_sac = np.sum(prb_utilization_sac[e, :, t])
        sum_of_squares_sac = np.sum(prb_utilization_sac[e, :, t] ** 2)


        sum_of_prbs_heuristic = np.sum(prb_utilization_heuristic[e, :, t])
        sum_of_squares_heuristic = np.sum(prb_utilization_heuristic[e, :, t] ** 2)

        sum_of_prbs_ddpg = np.sum(prb_utilization_ddpg[e, :, t])
        sum_of_squares_ddpg = np.sum(prb_utilization_ddpg[e, :, t] ** 2)

        # Handle potential division by zero or NaN
        if sum_of_prbs_sac == 0 or np.isnan(sum_of_squares_sac):
            fairness_scores_sac[e, t] = 0  # Set fairness score to 0
        else:
            fairness_scores_sac[e, t] = (sum_of_prbs_sac ** 2) / (BS_NO * sum_of_squares_sac)
        ##
        if sum_of_prbs_heuristic == 0 or np.isnan(sum_of_squares_heuristic):
            fairness_scores_heuristic[e, t] = 0
        else:
            fairness_scores_heuristic[e, t] = (sum_of_prbs_heuristic ** 2) / (BS_NO * sum_of_squares_heuristic)

        
        if sum_of_prbs_ddpg == 0 or np.isnan(sum_of_squares_ddpg):
            fairness_scores_ddpg[e, t] = 0  # Set fairness score to 0
        else:
            fairness_scores_ddpg[e, t] = (sum_of_prbs_ddpg ** 2) / (BS_NO * sum_of_squares_ddpg)
        ##

##Average fairness scores over all MC runs
# avg_fairness_scores = np.mean(fairness_scores, axis=0)
avg_fairness_scores_sac = np.mean(fairness_scores_sac, axis=0)
normalized_fairness_scores_sac = (avg_fairness_scores_sac - np.min(avg_fairness_scores_sac)) / (np.max(avg_fairness_scores_sac) - np.min(avg_fairness_scores_sac))

avg_fairness_scores_heuristic = np.mean(fairness_scores_heuristic, axis=0)
normalized_fairness_scores_heuristic = (avg_fairness_scores_heuristic - np.min(avg_fairness_scores_heuristic)) / (np.max(avg_fairness_scores_heuristic) - np.min(avg_fairness_scores_heuristic))

avg_fairness_scores_ddpg = np.mean(fairness_scores_ddpg, axis=0)
normalized_fairness_scores_ddpg = (avg_fairness_scores_ddpg - np.min(avg_fairness_scores_ddpg)) / (np.max(avg_fairness_scores_ddpg) - np.min(avg_fairness_scores_ddpg))


data = [moving_average(normalized_fairness_scores_ddpg, WINDOW_SIZE), moving_average(normalized_fairness_scores_sac, WINDOW_SIZE), moving_average(normalized_fairness_scores_heuristic, WINDOW_SIZE)]
names = ['DDPG', 'SAC','Heuristic']
averages = [np.average(normalized_fairness_scores_ddpg),np.average(normalized_fairness_scores_sac), np.average(normalized_fairness_scores_heuristic)]
labels = ['{}: {:.3f} (on average)'.format(n, a) for n, a in zip(names, averages)]

colors = ['b', 'g', 'r']  # choose colors for each curve
linestyles = ['-', '-', '-']  # choose line styles for each curve

plot_graph('Normalized Jain\'s fairness score', data, labels, colors, linestyles, "Episode", "Fairness score")
```
