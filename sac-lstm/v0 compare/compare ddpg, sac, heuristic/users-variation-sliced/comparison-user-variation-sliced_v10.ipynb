```python
#SSL_u JFI
## Calculate proportional fairness score for individual bitrate for each Episode
E=1
T=4200
USER_NO=10
WINDOW_SIZE=300

# Function to calculate Jain's Fairness Index
def calculate_jfi(values):
    numerator = np.sum(values) ** 2
    denominator = len(values) * np.sum(values ** 2)
    if denominator == 0:
        return 0  # Avoid division by zero
    return numerator / denominator

# Calculate JFI for each episode and each time slot
jfi_values_sac_10u = np.zeros((E, T))
jfi_values_ddpg_10u = np.zeros((E, T))
jfi_values_heuristic_10u = np.zeros((E, T))
for e in range(E):
    for t in range(T):
        rate_t_sac_10u = rate_sac_10u[e, :, t]
        jfi_values_sac_10u[e, t] = calculate_jfi(rate_t_sac_10u)

        rate_t_ddpg_10u = rate_ddpg_10u[e, :, t]
        jfi_values_ddpg_10u[e, t] = calculate_jfi(rate_t_ddpg_10u)

        rate_t_heuristic_10u = rate_heuristic_10u[e, :, t]
        jfi_values_heuristic_10u[e, t] = calculate_jfi(rate_t_heuristic_10u)

# Average the JFI values over all episodes for each time slot
average_jfi_values_sac_10u = np.mean(jfi_values_sac_10u, axis=0)
average_jfi_values_ddpg_10u = np.mean(jfi_values_ddpg_10u, axis=0)
average_jfi_values_heuristic_10u = np.mean(jfi_values_heuristic_10u, axis=0)

data = [moving_average(average_jfi_values_ddpg_10u, WINDOW_SIZE), moving_average(average_jfi_values_sac_10u, WINDOW_SIZE), moving_average(average_jfi_values_heuristic_10u, WINDOW_SIZE)]
names = ['DDPG', 'SAC','Heuristic']
averages = [np.average(average_jfi_values_ddpg_10u),np.average(average_jfi_values_sac_10u), np.average(average_jfi_values_heuristic_10u)]
labels = ['{}: {:.2f} (on average)'.format(n, a) for n, a in zip(names, averages)]

colors = ['b', 'g', 'r']  # choose colors for each curve
linestyles = ['-', '-', '-']  # choose line styles for each curve

plot_graph('Normalized Jain\'s fairness score', data, labels, colors, linestyles, "Episode", "Fairness score")
```

```python
############
## Calculate PRB utilization for each user at each Episode and for each MC run
prb_utilization_sac_10u = np.sum(mat_rho_sac_10u, axis=(0, 1))  # Sum over PRBs and BSs
prb_utilization_heuristic_10u = np.sum(mat_rho_heuristic_10u, axis=(0, 1))  # Sum over PRBs and BSs
prb_utilization_ddpg_10u = np.sum(mat_rho_ddpg_10u, axis=(0, 1))  # Sum over PRBs and BSs

## Calculate proportional fairness score for each Episode
E=1
T=5000
BS_NO=4
fairness_scores_sac_10u = np.zeros((E, T))
fairness_scores_heuristic_10u = np.zeros((E, T))
fairness_scores_ddpg_10u = np.zeros((E, T))

for e in range(E):
    for t in range(T-1):
        # Calculate Jain's fairness index for Episode t in MC run mc
        sum_of_prbs_sac_10u = np.sum(prb_utilization_sac_10u[e, :, t])
        sum_of_squares_sac_10u = np.sum(prb_utilization_sac_10u[e, :, t] ** 2)


        sum_of_prbs_heuristic_10u = np.sum(prb_utilization_heuristic_10u[e, :, t])
        sum_of_squares_heuristic_10u = np.sum(prb_utilization_heuristic_10u[e, :, t] ** 2)

        sum_of_prbs_ddpg_10u = np.sum(prb_utilization_ddpg_10u[e, :, t])
        sum_of_squares_ddpg_10u = np.sum(prb_utilization_ddpg_10u[e, :, t] ** 2)

        # Handle potential division by zero or NaN
        if sum_of_prbs_sac_10u == 0 or np.isnan(sum_of_squares_sac_10u):
            fairness_scores_sac_10u[e, t] = 0  # Set fairness score to 0
        else:
            fairness_scores_sac_10u[e, t] = (sum_of_prbs_sac_10u ** 2) / (BS_NO * sum_of_squares_sac_10u)
        ##
        if sum_of_prbs_heuristic_10u == 0 or np.isnan(sum_of_squares_heuristic_10u):
            fairness_scores_heuristic_10u[e, t] = 0
        else:
            fairness_scores_heuristic_10u[e, t] = (sum_of_prbs_heuristic_10u ** 2) / (BS_NO * sum_of_squares_heuristic_10u)

        
        if sum_of_prbs_ddpg_10u == 0 or np.isnan(sum_of_squares_ddpg_10u):
            fairness_scores_ddpg_10u[e, t] = 0  # Set fairness score to 0
        else:
            fairness_scores_ddpg_10u[e, t] = (sum_of_prbs_ddpg_10u ** 2) / (BS_NO * sum_of_squares_ddpg_10u)
        ##

##Average fairness scores over all MC runs
# avg_fairness_scores = np.mean(fairness_scores, axis=0)
avg_fairness_scores_sac_10u = np.mean(fairness_scores_sac_10u, axis=0)
normalized_fairness_scores_sac_10u = (avg_fairness_scores_sac_10u - np.min(avg_fairness_scores_sac_10u)) / (np.max(avg_fairness_scores_sac_10u) - np.min(avg_fairness_scores_sac_10u))

avg_fairness_scores_heuristic_10u = np.mean(fairness_scores_heuristic_10u, axis=0)
normalized_fairness_scores_heuristic_10u = (avg_fairness_scores_heuristic_10u - np.min(avg_fairness_scores_heuristic_10u)) / (np.max(avg_fairness_scores_heuristic_10u) - np.min(avg_fairness_scores_heuristic_10u))

avg_fairness_scores_ddpg_10u = np.mean(fairness_scores_ddpg_10u, axis=0)
normalized_fairness_scores_ddpg_10u = (avg_fairness_scores_ddpg_10u - np.min(avg_fairness_scores_ddpg_10u)) / (np.max(avg_fairness_scores_ddpg_10u) - np.min(avg_fairness_scores_ddpg_10u))


data = [moving_average(normalized_fairness_scores_ddpg_10u, WINDOW_SIZE), moving_average(normalized_fairness_scores_sac_10u, WINDOW_SIZE), moving_average(normalized_fairness_scores_heuristic_10u, WINDOW_SIZE)]
names = ['DDPG', 'SAC','Heuristic']
averages = [np.average(normalized_fairness_scores_ddpg_10u),np.average(normalized_fairness_scores_sac_10u), np.average(normalized_fairness_scores_heuristic_10u)]
labels = ['{}: {:.3f} (on average)'.format(n, a) for n, a in zip(names, averages)]

colors = ['b', 'g', 'r']  # choose colors for each curve
linestyles = ['-', '-', '-']  # choose line styles for each curve

plot_graph('Normalized Jain\'s fairness score', data, labels, colors, linestyles, "Episode", "Fairness score")
```
