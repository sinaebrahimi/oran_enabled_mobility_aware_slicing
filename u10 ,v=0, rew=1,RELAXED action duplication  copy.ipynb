```python
import numpy as np

# Assume we have USER_NO users, BS_NO base stations, and each BS has PRB_NO PRBs
USER_NO = 10
BS_NO = 3
PRB_NO = 5
MAX_POWER = 100  # example max power for each BS
self_SIGMA_NOISE = 1e-9  # example noise power

# Example channel gains (random for illustration)
h = np.random.rand(BS_NO, PRB_NO, USER_NO)

# Binary allocation variables (random for illustration)
rho = np.random.randint(0, 2, (BS_NO, PRB_NO, USER_NO))
chi = np.random.randint(0, 2, (USER_NO, BS_NO))

# DRL output for each user's normalized power allocation
self_temp_p_reshaped = np.random.rand(USER_NO)

# Calculate inverse SINR
inverse_sinr = self_SIGMA_NOISE / h

# Initialize power allocation matrix
P = np.zeros((BS_NO, PRB_NO, USER_NO))

# Iterate over each BS
for b in range(BS_NO):
    # Get users assigned to BS b
    users_assigned = np.where(chi[:, b] == 1)[0]
    
    if len(users_assigned) > 0:
        # Normalize DRL output for users assigned to BS b
        normalized_p = self_temp_p_reshaped[users_assigned] / np.sum(self_temp_p_reshaped[users_assigned])
        
        # Power allocated to each user in BS b
        user_power_allocation = MAX_POWER * normalized_p
        
        # Allocate power to each PRB based on inverse SINR and binary variables
        for u in users_assigned:
            user_power = user_power_allocation[users_assigned.tolist().index(u)]
            
            # Sort PRBs based on inverse SINR for user u in BS b
            prb_indices = np.where(rho[b, :, u] == 1)[0]
            sorted_prbs = prb_indices[np.argsort(inverse_sinr[b, prb_indices, u])]
            
            # Water filling allocation
            remaining_power = user_power
            for k in sorted_prbs:
                if remaining_power <= 0:
                    break
                allocated_power = max(0, remaining_power - inverse_sinr[b, k, u])
                P[b, k, u] = allocated_power
                remaining_power -= allocated_power
    else:
        # No users assigned to this BS, so no power allocation needed
        pass

# Print allocated power matrix
print(P)
```
